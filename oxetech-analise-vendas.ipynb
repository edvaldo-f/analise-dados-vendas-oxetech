Percebe dois pontos de atenção:


*   Não conseguimos verificar todos os valores de valor_total, para perceber se possui erros em sua escrita
*   Há um desconto com valor 1.20, o que está incorreto, mas qual seria o valor de desconto então? Vamos ver mais adiante.



```python
# 1. Define o padrão: apenas números inteiros ou decimais com ponto.
padrao = r'^\d+(\.\d+)?$'

# 2. Identifica o que não bate com o padrão.
mascara_validos = df['valor_total'].astype(str).str.match(padrao, na=False)
erros = df[~mascara_validos]

# 3. Diagnóstico.
print(f"Total de registros fora do padrão: {len(erros)}")

if not erros.empty:
    print("\nAmostra dos valores inválidos encontrados (valores únicos):")
    display(erros['valor_total'].unique()[:20])

    print("\nVisualização das primeiras linhas completas com erro:")
    display(erros[['valor_total']].head())
else:
    print("Nenhum erro encontrado! Todos os dados estão no formato correto.")
```

Percebemos que o problema ocorre por valores nulos, os quais serão tratados mais adiante.

Partiremos para avaliar os valores '1.20' em desconto_pct.

Podemos prezumir que são valores errados, pois os descontos sobem de forma gradual e baixa, e um desconto de 120% geraria um valor_total negativo.

Sendo assim, devemos corrigir.

```python
# Visualizando as linhas com desconto errado...
linhas_com_desconto_errado = df[df["desconto_pct"] == 1.20]

linhas_com_desconto_errado.head()
```

Para garantir que o valor seja o correto, podemos verificar a fórmula utilizada para calcular o valor_total:

`
valor_total = (preco_unitario * quantidade) * (1 - desconto_pct)
`

Ou melhor, uma variação dela:

`
desconto = 1 - (valor_total / (quantidade * preco_unitario))
`

```python
# Criando outra coluna temporária para comparação

df["desconto_esperado"] = (
    1 - (df["valor_total"] / (df["quantidade"] * df["preco_unitario"]))
).round(2)
```

```python
# Comparando com os valores errados...

df.loc[
    df["desconto_pct"] == 1.20,
    ["desconto_pct", "desconto_esperado"]
]
```

Com os valores corretos encontrados, podemos substituí-los, da mesma forma anteriormente feita.

```python
# Correção dos descontos errados...

descontos_alvo = df["desconto_pct"] == 1.20

df.loc[descontos_alvo, "desconto_pct"] = df["desconto_esperado"]
```

```python
# Exibição novamente

df["desconto_pct"].value_counts()
```

Com todos os descontos corrigidos agora, podemos deletar a coluna temporária e seguir.

Alguns descontos antes podiam aparecer com valores errados, pelo valor_total estar incorreto, mas essa coluna será trata mais adiante.

```python
# Exclusão da coluna temporária

df = df.drop(columns= ["desconto_esperado"])
```

### 5.2 Valores mais afundo
Neste tópico fazemos análises mais afundo, com conhecimento mais estatístico.
Quartis, IQR.

Exibiremos os quartis 1,2 e 3.

Aos quais já foram encontrados no .describe(), mas vamos colocá-los de forma mais dedicada, e depois encontrar o IQR com eles.

```python
# Calculando e exibindo os quartis...
q1 = df["valor_total"].quantile(0.25)
q2 = df["valor_total"].quantile(0.50)
q3 = df["valor_total"].quantile(0.75)

print(f"1° Quartil: {q1:.2f}")
print(f"2° Quartil: {q2:.2f}")
print(f"3° Quartil: {q3:.2f}")
```

Os resultados correspondem corretamente.

Reforçando que o 2° quartil corresponde a Mediana.

Além disso, analisando os quartis agora, é perceptível uma variação leve entre ele, os "saltos" entre os dados não são gritantes. Oque também confirma que o valor de 1249,50 é um outlier.

Analisaremos o IQR agora, juntamente aos limites superior e inferior.

A distância entre o 1° quartil e o 3° quartil.

```python
# Calculando o IQR...

iqr = q3 - q1

# Calculando os limites...

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

# Exibindo ...

print(f"O IQR é: {iqr:.2f}")
print(f"O limite inferior é: {limite_inferior:.2f}")
print(f"O limite superior é: {limite_superior:.2f}")
```

Esses dados nos ajudam a encontrar outliers, que é oque faremos logo agora.

```python
# Encontrando os outliers...

outliers = df[
    (df["valor_total"] < limite_inferior) |
    (df["valor_total"] > limite_superior)
]

print(f"Quantidade de outliers: {len(outliers)}")
```

```python
# Visualizando mais diretamente...

outliers[["quantidade", "produto", "preco_unitario", "valor_total"]].sort_values(by="valor_total", ascending=False)
```

Através da céçula acima descobrimos os outliers, além do seu valor gasto, quantidade de produtos comprados, o produto em questão e o valor do produto.

A maior parte das compras foram de tênis com quantidades 5 sendo a maioria, exceto por compras de quantidade 4 ou a compra de 5 calças jeans.
